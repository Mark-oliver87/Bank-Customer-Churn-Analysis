# Bank Customer Churn Analysis

## Project Overview

Customer churn is an important challenge in the banking industry because retaining existing customers can be critical to maintaining revenue and long-term customer relationships.

This project analyzes bank customer data to identify patterns associated with customer churn and to understand which customer characteristics are linked to a higher likelihood of leaving the bank.

The analysis combines exploratory data analysis, customer segmentation, visualization, and business-focused interpretation to generate insights that could support customer retention strategies.

## Business Questions

This analysis aims to answer the following questions:

1. What is the overall customer churn rate?
2. How does churn vary across customer demographics such as age, gender, and geography?
3. Are customers with higher account balances more likely to churn?
4. How do credit score and customer tenure relate to churn?
5. Does the number of bank products used by a customer influence churn?
6. Are inactive customers more likely to leave the bank?
7. Does having a credit card appear to be associated with customer retention?
8. Which customer segments show the highest churn rates?
9. What actionable strategies could the bank consider to improve customer retention?

## Exploratory Data Analysis

### Overall Customer Churn

The first step is to examine the overall proportion of customers who remained with the bank versus those who exited.

```python
df["Exited"].value_counts()
```

#### Key Finding

Out of 10,000 customers, 2,037 exited the bank, resulting in an overall churn rate of **20.37%**. The remaining **79.63%** of customers stayed.

This means that approximately one in five customers in the dataset churned, indicating that customer retention is a meaningful business issue for further investigation.

### Churn by Geography

Customer churn is compared across geographic markets to determine whether churn patterns differ by country.

```python
df.groupby("Geography")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

Churn varies substantially across the three geographic markets. Germany has the highest observed churn rate at **32.44%**, compared with **16.67%** in Spain and **16.15%** in France.

Although Germany has approximately half as many customers as France, it recorded slightly more churned customers (814 versus 810). This suggests that the German customer segment warrants further investigation to understand which customer characteristics may be associated with its higher churn rate.

### Churn by Gender

Churn rates are compared across gender groups to determine whether customer attrition differs between the groups represented in the dataset.

```python
df.groupby("Gender")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

The analysis shows that although there are slightly fewer female customers than male customers, female customers have a higher churn rate. Approximately **1 in 4 female customers (25.07%)** exited the bank, compared with **16.46%** of male customers.

This suggests that the higher churn observed among female customers warrants further investigation to identify other factors that may be associated with this difference.

### Churn by Age

Customer age is examined to determine whether customers who exited the bank differ in age from customers who remained.

```python
df.groupby("Exited")["Age"].agg(["count", "mean", "median", "min", "max"])
```

```python
age_bins = [18, 30, 40, 50, 60, 70, 100]
age_labels = ["18-29", "30-39", "40-49", "50-59", "60-69", "70+"]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels,
    right=False
)
```

```python
df[["Age", "AgeGroup"]].head(10)
```

```python
df.groupby("AgeGroup", observed=True)["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

Churn increases substantially across the younger and middle-aged groups, reaching its highest level among customers aged **50–59**, where approximately **56.04%** of customers exited the bank.

After this age group, the churn rate declines to **35.20%** among customers aged 60–69 and **9.93%** among customers aged 70+. The results therefore suggest that the relationship between age and churn is not simply linear, with customers aged **40–59** showing particularly elevated churn in this dataset.

The relatively small number of customers in the oldest age groups should also be considered when interpreting these results.

### Churn and Account Balance

Account balances are compared between customers who remained with the bank and those who exited to determine whether balance levels are associated with customer churn.

```python
df.groupby("Exited")["Balance"].agg(["count", "mean", "median", "min", "max"])
```

```python
df.groupby("Exited")["Balance"].apply(lambda x: (x == 0).mean() * 100)
```

```python
def balance_group(balance):
    if balance == 0:
        return "Zero Balance"
    elif balance < 50000:
        return "Below 50K"
    elif balance < 100000:
        return "50K-99K"
    elif balance < 150000:
        return "100K-149K"
    else:
        return "150K+"

df["BalanceGroup"] = df["Balance"].apply(balance_group)
```

```python
df[["Balance", "BalanceGroup"]].head(10)
```

```python
df.groupby("BalanceGroup")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

Customers with a zero balance had a relatively low churn rate of **13.82%**, while customers with balances between **100K and 149K** had a higher churn rate of **25.77%**.

The Below 50K group recorded the highest churn rate at **34.67%**, but this segment contains only **75 customers**, so the result should be interpreted cautiously.

Overall, customers carrying substantial account balances show higher observed churn rates than zero-balance customers in this dataset, although the analysis does not establish that higher balances cause churn.

### Churn and Credit Score

Credit scores are compared between customers who stayed and customers who exited to determine whether creditworthiness is associated with customer churn.

```python
df.groupby("Exited")["CreditScore"].agg(["count", "mean", "median", "min", "max"])
```

```python
credit_bins = [300, 580, 670, 740, 800, 851]
credit_labels = ["Poor", "Fair", "Good", "Very Good", "Excellent"]

df["CreditScoreGroup"] = pd.cut(
    df["CreditScore"],
    bins=credit_bins,
    labels=credit_labels,
    right=False
)
```

```python
df[["CreditScore", "CreditScoreGroup"]].head(10)
```

```python
df.groupby("CreditScoreGroup", observed=True)["Exited"].agg(
    ["count", "sum", "mean"]
)
```

#### Key Finding

Credit score shows relatively modest differences in churn across customer groups. Customers in the Poor credit-score group recorded the highest observed churn rate at **22.02%**, while the Good group recorded the lowest at **18.62%**.

However, churn does not consistently decrease as credit scores improve. Combined with the relatively small difference in average credit scores between customers who stayed and those who exited, this suggests that **credit score alone is not a strong differentiator of churn in this dataset**.

### Churn by Customer Tenure

Customer tenure is examined to determine whether the length of the banking relationship is associated with customer churn.

```python
df.groupby("Tenure")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

Customers with **0 years of tenure** recorded the highest observed churn rate at approximately **23.00%**, followed by customers with 1 year of tenure at **22.42%**. Customers with 7 years of tenure recorded the lowest churn rate at **17.22%**.

However, churn does not consistently increase or decrease as tenure rises. The rates fluctuate across tenure levels, suggesting that **tenure alone is not a strong differentiator of customer churn in this dataset**.

### Churn by Number of Products

The number of bank products held by each customer is examined to determine whether the depth of the customer relationship is associated with churn.

```python
df.groupby("NumOfProducts")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

The number of products held by customers shows a strong but non-linear relationship with churn. Customers with **2 products** recorded the lowest churn rate at **7.58%**, compared with **27.71%** among customers holding only 1 product.

Churn rises sharply among customers with 3 products (**82.71%**) and 4 products (**100%**). However, these groups are relatively small, with only 266 and 60 customers respectively, so these results should be interpreted cautiously.

The unusually high churn among customers holding 3–4 products warrants further investigation, while the low churn among customers with 2 products suggests that product engagement may be associated with customer retention.

### Churn by Customer Activity

Customer activity status is examined to determine whether active engagement with the bank is associated with customer retention.

```python
df.groupby("IsActiveMember")["Exited"].agg(["count", "sum", "mean"])
```

#### Key Finding

Inactive customers recorded a substantially higher churn rate of **26.85%**, compared with **14.27%** among active customers.

This indicates a strong association between customer activity and retention in this dataset. Inactive customers may therefore represent an important segment for further investigation and potential re-engagement or retention initiatives.

### Churn by Credit Card Ownership

Credit card ownership is examined to determine whether customers who hold a bank credit card have different churn rates from customers who do not.

```python
df.groupby("HasCrCard")["Exited"].agg(["count", "sum", "mean"])
```

### Churn and Estimated Salary

Estimated salaries are compared between customers who stayed and customers who exited to determine whether income level is associated with customer churn.

```python
df.groupby("Exited")["EstimatedSalary"].agg(
    ["count", "mean", "median", "min", "max"]
)
```

```python
salary_bins = [0, 50000, 100000, 150000, 200001]
salary_labels = ["Below 50K", "50K-99K", "100K-149K", "150K+"]

df["SalaryGroup"] = pd.cut(
    df["EstimatedSalary"],
    bins=salary_bins,
    labels=salary_labels,
    right=False
)
```

```python
df[["EstimatedSalary", "SalaryGroup"]].head(10)
```

```python
df.groupby("SalaryGroup", observed=True)["Exited"].agg(
    ["count", "sum", "mean"]
)
```

#### Key Finding

Estimated salary shows relatively little variation in churn across income groups. Churn rates range from approximately **19.87% to 21.47%**, with the 150K+ group recording the highest observed rate.

Combined with the similar average and median salaries of customers who stayed and those who exited, this suggests that **estimated salary alone is not a strong differentiator of customer churn in this dataset**.

## Customer Segmentation Analysis

Single-variable analysis identified notable churn differences across geography, age, number of products, customer activity, and account balance. The next stage combines selected characteristics to identify customer segments with particularly high or low observed churn rates.

### Geography and Customer Activity

Geography and customer activity are analyzed together to determine whether churn differences across geographic markets vary according to customer engagement.

```python
geo_activity = (
    df.groupby(["Geography", "IsActiveMember"])["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

geo_activity
```

#### Key Finding

Combining geography and customer activity reveals substantial differences in churn. **Inactive customers in Germany recorded the highest churn rate at 41.08%**, compared with 21.13% among inactive customers in France and 23.35% among inactive customers in Spain.

Active customers recorded lower churn than inactive customers within all three geographic markets. However, active German customers still had a relatively elevated churn rate of **23.72%**, suggesting that Germany's higher observed churn cannot be explained by customer inactivity alone.

This indicates that customer retention strategies may benefit from considering both geographic market and customer engagement rather than examining either characteristic in isolation.

```python
age_activity = (
    df.groupby(["AgeGroup", "IsActiveMember"], observed=True)["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

age_activity
```

#### Key Finding

Combining age and customer activity reveals substantial differences in churn. Among customers aged **50–59**, inactive customers recorded a churn rate of **81.23%**, compared with **37.10%** among active customers.

The most extreme observed rate occurred among inactive customers aged **60–69**, at **91.49%**. However, this segment contains only 94 customers, so the result should be interpreted cautiously.

Overall, active customers recorded lower churn than inactive customers across every age group. The difference is particularly pronounced among customers aged 50–69, suggesting that customer engagement may be especially relevant when examining churn within these age segments.

```python
product_activity = (
    df.groupby(["NumOfProducts", "IsActiveMember"])["Exited"]
      .agg(["count", "sum", "mean"])
      .reset_index()
)

product_activity
```

#### Key Finding

Customer activity is associated with lower churn among customers holding 1 or 2 products. Among customers with 1 product, churn falls from **36.65% for inactive customers** to **18.92% for active customers**. Among customers with 2 products, churn falls from **9.89%** to **5.56%**.

However, customers holding 3 products recorded very high churn regardless of activity status: **88.24% among inactive customers** and **75.22% among active customers**. All customers holding 4 products churned in this dataset, although this group contains only 60 customers.

These results suggest that customer inactivity does not fully explain the unusually high churn observed among customers holding 3–4 products. This segment warrants further investigation into factors such as product suitability, pricing, service experience, or other characteristics not captured in the dataset.

## Key Insights and Business Recommendations

### Key Insights

The analysis identified several important patterns associated with customer churn:

1. **Overall churn is substantial:** 2,037 of 10,000 customers exited, representing an overall churn rate of **20.37%**.

2. **Geography is an important differentiator:** Customers in Germany recorded a churn rate of approximately **32.44%**, nearly double the rates observed in France and Spain.

3. **Age shows a strong non-linear relationship with churn:** Churn increases considerably among customers aged 40–59 and peaks at approximately **56.0%** for customers aged 50–59.

4. **Customer activity is strongly associated with retention:** Inactive customers recorded a churn rate of **26.85%**, compared with **14.27%** among active customers.

5. **Product holdings show a significant but non-linear pattern:** Customers holding 2 products recorded the lowest churn rate at **7.58%**, while customers with 3–4 products experienced substantially higher churn. The smaller size of these latter groups should be considered when interpreting the results.

6. **Customer segmentation reveals additional risk patterns:** Inactive German customers recorded a churn rate of approximately **41.1%**. Activity status was associated with lower churn across all three geographic markets.

7. **Several variables showed weaker standalone relationships with churn:** Credit score, tenure, credit card ownership, and estimated salary showed relatively small or inconsistent differences in churn rates.

### Business Recommendations

- **Prioritize customer re-engagement:** Develop targeted retention initiatives for inactive customers, particularly within higher-churn segments.

- **Investigate the German customer experience:** Examine service quality, product suitability, pricing, customer complaints, and other market-specific factors that may help explain the elevated churn observed in Germany.

- **Focus retention analysis on middle-aged customers:** Customers aged 40–59, particularly those aged 50–59, represent an important segment for further investigation and targeted retention efforts.

- **Review customers holding 3–4 products:** The unusually high churn within these groups warrants deeper investigation into product combinations, customer experience, fees, or other factors not captured in the current dataset.

- **Use segmentation rather than single-variable rules:** Combining characteristics such as geography, age, product holdings, and activity status provides a more useful view of customer churn risk than examining individual variables in isolation.

### Analytical Limitation

The findings represent associations observed in this dataset and should not be interpreted as evidence that these characteristics directly cause customer churn. Additional customer behavior, service, transaction, pricing, and satisfaction data would be valuable for understanding the underlying drivers.

## Predictive Modelling

The exploratory analysis identified several customer characteristics associated with churn. The next stage develops a classification model to determine whether customer information can be used to predict customer churn.

The target variable is **Exited**, where:

- `0` = Customer stayed
- `1` = Customer churned

Model performance will be evaluated on previously unseen test data to assess how well the model generalizes beyond the customers used during training.

### Feature Selection

Customer identifiers and administrative fields are excluded from modelling because they do not represent meaningful customer characteristics for predicting churn.

The model uses the following features:

- Credit Score
- Geography
- Gender
- Age
- Tenure
- Account Balance
- Number of Products
- Credit Card Ownership
- Active Membership Status
- Estimated Salary

```python
features = [
    "CreditScore",
    "Geography",
    "Gender",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "IsActiveMember",
    "EstimatedSalary"
]

X = df[features].copy()
y = df["Exited"].copy()
```

```python
X.shape, y.shape
```

```python
X.head()
```

### Train-Test Split

The dataset is divided into training and testing sets. The training data is used to develop the model, while the test data is kept separate and used to evaluate performance on unseen customers.

An 80/20 split is used, with stratification to preserve the proportion of churned and non-churned customers in both datasets.

```python
%pip install scikit-learn
```

```python
import sklearn
print(sklearn.__version__)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

### Data Preprocessing

The categorical variables `Geography` and `Gender` are converted into numerical features using one-hot encoding. Numerical variables are retained in their original form.

Preprocessing is fitted using the training data and then applied to the test data to avoid information leakage from the test set.

### Model 1: Logistic Regression

Logistic Regression is used as the baseline classification model. It provides a relatively simple and interpretable benchmark against which a more flexible model can later be compared.

The preprocessing and classification steps are combined in a pipeline so that categorical encoding is applied consistently during both training and prediction.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_features = ["Geography", "Gender"]

numeric_features = [
    "CreditScore",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "IsActiveMember",
    "EstimatedSalary"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(drop="first", handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            StandardScaler(),
            numeric_features
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)

logistic_model.fit(X_train, y_train)
```

```python
logistic_predictions = logistic_model.predict(X_test)

logistic_probabilities = logistic_model.predict_proba(X_test)[:, 1]
```

```python
logistic_predictions[:10]
```

### Logistic Regression Evaluation

The model is evaluated using accuracy, precision, recall, F1-score, ROC-AUC, and a confusion matrix. Because the dataset contains considerably fewer churned customers than retained customers, accuracy alone is not sufficient to assess model performance.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:", round(accuracy_score(y_test, logistic_predictions), 3))
print("Precision:", round(precision_score(y_test, logistic_predictions), 3))
print("Recall:", round(recall_score(y_test, logistic_predictions), 3))
print("F1 Score:", round(f1_score(y_test, logistic_predictions), 3))
print("ROC-AUC:", round(roc_auc_score(y_test, logistic_probabilities), 3))
```

```python
print(classification_report(
    y_test,
    logistic_predictions,
    target_names=["Stayed", "Churned"]
))
```

### Model 2: Random Forest

A Random Forest classifier is developed as a second model. Unlike Logistic Regression, Random Forest can capture more complex and non-linear relationships between customer characteristics and churn.

The model is evaluated on the same test dataset to allow a consistent comparison with the baseline model.

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

random_forest_model.fit(X_train, y_train)
```

```python
rf_predictions = random_forest_model.predict(X_test)
rf_probabilities = random_forest_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", round(accuracy_score(y_test, rf_predictions), 3))
print("Precision:", round(precision_score(y_test, rf_predictions), 3))
print("Recall:", round(recall_score(y_test, rf_predictions), 3))
print("F1 Score:", round(f1_score(y_test, rf_predictions), 3))
print("ROC-AUC:", round(roc_auc_score(y_test, rf_probabilities), 3))
```

```python
print(classification_report(
    y_test,
    rf_predictions,
    target_names=["Stayed", "Churned"]
))
```

### Model Comparison

```python
model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, logistic_predictions),
        accuracy_score(y_test, rf_predictions)
    ],
    "Precision": [
        precision_score(y_test, logistic_predictions),
        precision_score(y_test, rf_predictions)
    ],
    "Recall": [
        recall_score(y_test, logistic_predictions),
        recall_score(y_test, rf_predictions)
    ],
    "F1 Score": [
        f1_score(y_test, logistic_predictions),
        f1_score(y_test, rf_predictions)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, logistic_probabilities),
        roc_auc_score(y_test, rf_probabilities)
    ]
})

model_comparison.round(3)
```

**Interpretation:** Random Forest achieved stronger overall discrimination and higher precision and F1-score than Logistic Regression, with a ROC-AUC of approximately 0.85. Logistic Regression, however, achieved higher recall for churned customers (approximately 70% versus 60%). This demonstrates an important operational trade-off: a retention program prioritizing identification of as many potential churners as possible may prefer a higher-recall threshold or model, while a program seeking fewer false-positive interventions may place greater emphasis on precision and overall discrimination.

### Random Forest Feature Importance

Feature importance is examined to identify which customer characteristics contributed most to the Random Forest's predictions. These importance values indicate predictive contribution within the model and should not be interpreted as evidence of causation.

```python
feature_names = (
    random_forest_model
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

importances = (
    random_forest_model
    .named_steps["classifier"]
    .feature_importances_
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

feature_importance
```

```python
top_features = feature_importance.head(10).sort_values("Importance")

plt.figure(figsize=(9, 6))
plt.barh(top_features["Feature"], top_features["Importance"])

plt.title("Top Random Forest Feature Importances")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

## Predictive Modelling Summary

Two classification models were developed to predict customer churn: Logistic Regression and Random Forest.

### Model Performance

Logistic Regression achieved approximately **71% accuracy** and identified about **70% of actual churners**, demonstrating relatively strong recall. However, its churn precision was approximately **39%**, indicating a relatively high number of false-positive churn predictions.

Random Forest achieved approximately **85% accuracy**, **63% churn precision**, **60% churn recall**, an F1-score of approximately **0.61**, and a ROC-AUC of approximately **0.85**. It therefore demonstrated stronger overall discrimination and a better balance between precision and recall at the default classification threshold.

The comparison also highlights an important business trade-off. Logistic Regression identified a larger proportion of actual churners, while Random Forest produced fewer false-positive churn alerts and stronger overall predictive performance.

### Business Application

Rather than treating model predictions as automatic decisions, churn probabilities could be used to support customer-retention prioritization. Customers with higher predicted churn probabilities could be reviewed alongside customer value, engagement history, product holdings, service interactions, and the cost of retention interventions.

The appropriate classification threshold should ultimately depend on the business objective and the relative cost of failing to identify a customer who churns versus unnecessarily targeting a customer who would have remained.

### Limitations

The models are based only on the variables available in the current dataset. Additional information such as transaction activity, complaints, digital engagement, service interactions, product usage, fees, and customer satisfaction could potentially improve predictive performance.

Feature importance represents predictive contribution within the fitted model and does not establish that a characteristic causes customer churn. Model performance should also be validated on new customer data before operational deployment.

```python
import sqlite3
import pandas as pd

# Load the original customer dataset
sql_df = pd.read_csv("../data/Churn_Modelling.csv")

# Create/connect to SQLite database
conn = sqlite3.connect("../sql/bank_churn.db")

# Store the dataset as a SQL table
sql_df.to_sql(
    "customers",
    conn,
    if_exists="replace",
    index=False
)

conn.close()

print("SQLite database created successfully.")
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../sql/bank_churn.db")

query = """
SELECT
    COUNT(*) AS total_customers
FROM customers;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Geography,
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers
GROUP BY Geography
ORDER BY churn_rate_pct DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    CASE
        WHEN IsActiveMember = 1 THEN 'Active'
        ELSE 'Inactive'
    END AS activity_status,
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers
GROUP BY IsActiveMember
ORDER BY churn_rate_pct DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    CASE
        WHEN Age BETWEEN 18 AND 29 THEN '18-29'
        WHEN Age BETWEEN 30 AND 39 THEN '30-39'
        WHEN Age BETWEEN 40 AND 49 THEN '40-49'
        WHEN Age BETWEEN 50 AND 59 THEN '50-59'
        WHEN Age BETWEEN 60 AND 69 THEN '60-69'
        ELSE '70+'
    END AS age_group,
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers
GROUP BY age_group
ORDER BY
    CASE age_group
        WHEN '18-29' THEN 1
        WHEN '30-39' THEN 2
        WHEN '40-49' THEN 3
        WHEN '50-59' THEN 4
        WHEN '60-69' THEN 5
        WHEN '70+' THEN 6
    END;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    NumOfProducts AS number_of_products,
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers
GROUP BY NumOfProducts
ORDER BY NumOfProducts;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Geography,
    CASE
        WHEN IsActiveMember = 1 THEN 'Active'
        ELSE 'Inactive'
    END AS activity_status,
    CASE
        WHEN Age BETWEEN 18 AND 29 THEN '18-29'
        WHEN Age BETWEEN 30 AND 39 THEN '30-39'
        WHEN Age BETWEEN 40 AND 49 THEN '40-49'
        WHEN Age BETWEEN 50 AND 59 THEN '50-59'
        WHEN Age BETWEEN 60 AND 69 THEN '60-69'
        ELSE '70+'
    END AS age_group,
    COUNT(*) AS total_customers,
    SUM(Exited) AS churned_customers,
    ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
FROM customers
GROUP BY
    Geography,
    IsActiveMember,
    age_group
HAVING COUNT(*) >= 100
ORDER BY churn_rate_pct DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH age_churn AS (
    SELECT
        Geography,
        CASE
            WHEN Age BETWEEN 18 AND 29 THEN '18-29'
            WHEN Age BETWEEN 30 AND 39 THEN '30-39'
            WHEN Age BETWEEN 40 AND 49 THEN '40-49'
            WHEN Age BETWEEN 50 AND 59 THEN '50-59'
            WHEN Age BETWEEN 60 AND 69 THEN '60-69'
            ELSE '70+'
        END AS age_group,
        COUNT(*) AS total_customers,
        SUM(Exited) AS churned_customers,
        ROUND(AVG(Exited) * 100, 2) AS churn_rate_pct
    FROM customers
    GROUP BY Geography, age_group
),

ranked_segments AS (
    SELECT
        *,
        RANK() OVER (
            PARTITION BY Geography
            ORDER BY churn_rate_pct DESC
        ) AS churn_rank
    FROM age_churn
)

SELECT *
FROM ranked_segments
ORDER BY Geography, churn_rank;
"""

pd.read_sql_query(query, conn)
```
